```python
# some global variables/dialogue
opened = "You have opened the refrigerator.\nThese are your current details.\n"
closed = "\nYou have closed the refrigerator."
options = "\n\tTo see everything, input 'All'.\n\tTo see items running low, input 'Restock'.\n\tTo close the fridge, input 'No'.\n"
search = "What would you like to look at?" + options + "\n"
additional = "\nWould you like to look at anything else?" + options
restock = "\nThese items are running low. You might consider going grocery shopping."

a = "Fridge info"
b = "Fridge contents"
c = "Freezer contents"
```

```python
import json

def print_all(x):
    #prints everything via (key, value)
    print(x)
    for key, value in refrigerator[x].items():
        if 'contents' in x:
            print(f"\t{key}: {value['quantity']} {value['unit']}")
        else:
            print(f"\t{key}: {value}")
    print("\n")

def new_input(x): 
    user = input(additional)
    return user

def get_restock(x,y):
    #prints out contents that need restocking
    print(restock)
    
    for key, value in refrigerator[x].items():
        if value['quantity'] <= 1:
            print(f"\t{key}: {value['quantity']} {value['unit']}")
    for key, value in refrigerator[y].items():
        if value['quantity'] <= 1:
            print(f"\t{key}: {value['quantity']} {value['unit']}")

    
#check if .json file exists; if yes, fridge is opened
try: 
    file = open("refrigerator_contents.json", "r")
    print(opened)
    
    #read .json file into a python dictionary
    #json.load() = reads a json file and returns a python object
    refrigerator = json.load(file)
    for item in refrigerator:
        print(item)
    print("\n")
except IOError: 
    print("Error: File not found") 

#allow end-user to issue a query and extract a subset of data
user = input(search)

while refrigerator:
    #input valid regardless of upper/lower case
    if user.lower() == "fridge info":
        print_all(a)
        user = new_input(user)
    elif user.lower() == "fridge contents":
        print_all(b)
        user = new_input(user)
    elif user.lower() == "freezer contents":
        print_all(c)
        user = new_input(user)
    elif user.lower() == "all":
        print_all(a)
        print_all(b)
        print_all(c)
        user = new_input(user)
    elif user.lower() == "restock":
        get_restock(b,c)
        user = new_input(user)
    elif user.lower() == "no":
        #close the file/fridge
        file.close()
        print(closed)
        break
    else:
        #if user enters item not in fridge or 2 items
        print("Details not found. Please ensure input is valid.\n" + search)
        user = input(search)
```
