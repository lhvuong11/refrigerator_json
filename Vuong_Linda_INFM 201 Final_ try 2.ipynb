```python
# some global variables/dialogue
opened = "You have opened the refrigerator.\nThese are your current details.\n"
closed = "\nYou have closed the refrigerator."
options = "\n\tTo see all grocery items, input 'All'.\n\tTo request specific items, input 'Retrieve [item]'.\n\tTo see which items to restock on, input 'Restock'.\n\tTo close the fridge, input 'No'.\n"
search = "What would you like to see?" + options + "\n"
additional = "\nWould you like to look at anything else?" + options
restock = "\nThese items are running low. You might consider going grocery shopping."
not_in = "\nThis item is either not in the fridge or doesn't exist. Consider buying some!\n"

a = "Fridge contents"
b = "Freezer contents"
```

```python
import json

def print_key(x):    #prints only key
    print(x)
    for key, value in refrigerator[x].items():
            print(f"\t{key}")
    print("\n")

def print_all(x):    #prints everything
    print(x)
    for key, value in refrigerator[x].items():
        print(f"\t{key}: {value['quantity']} {value['unit']}")
    print("\n")

def list_item(child):     #lists everything of that item
    child = child.capitalize()
            
    #finding the parent
    if child in refrigerator[a].keys():
        parent = a
    elif child in refrigerator[b].keys():
        parent = b
    else: 
        #item doesn't have a parent = not in the fridge
        print(not_in)
        return
        
    for key, value in refrigerator[parent].items():
        if key == child:
            print(f"\t{key}: {value['quantity']} {value['unit']}")
    print("\n")

def new_input(x): 
    user = input(additional)
    return user

def get_restock(x,y):
    #prints out contents that need restocking
    print(restock)
    
    for key, value in refrigerator[x].items():
        if value['quantity'] <= 1:
            print(f"\t{key}: {value['quantity']} {value['unit']}")
    for key, value in refrigerator[y].items():
        if value['quantity'] <= 1:
            print(f"\t{key}: {value['quantity']} {value['unit']}")

    
#check if .json file exists; if yes, fridge is opened
try: 
    file = open("refrigerator_contents.json", "r")
    print(opened)
    
    #read .json file into a python dictionary
    #json.load() = reads a json file and returns a python object
    refrigerator = json.load(file)
    for item in refrigerator:
        print(item)
    print("\n")
except IOError: 
    print("Error: File not found") 

#allow end-user to issue a query and extract a subset of data
user = input(search)

while refrigerator:
    #input valid regardless of upper/lower case
    
    if user.lower() == "all":
        print_key(a)
        print_key(b)
    elif user.lower() == "fridge contents":
        print_all(a)
    elif user.lower() == "freezer contents":
        print_all(b)
    elif "retrieve" in user.lower():
        #output that item and its quantity
        u = user.split()
        item = u[1]        #storing user item
        list_item(item)    #checking if item in fridge
    elif user.lower() == "restock":
        get_restock(a,b)
    elif user.lower() == "no":
        #close the file/fridge
        file.close()
        print(closed)
        break
    else:
        print("\nItem not found. Please ensure input is valid.\n")

    user = new_input(user)
```
