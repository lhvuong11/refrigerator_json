### Documentation
The purpose of this script is to allow the end-user to issue a query and extract a subset of data by reading in a .json file into a python dictionary. 
Based on the end-user's choices, it will output a .json file consisting of select keys.

### Version:
Python 3.13.5

### Vision/Scenario
For this scenario, the end-user is trying to examine the contents of their refrigerator. The end-user does this by issuing a query and depending on the data extracted, they might need to create a grocery list and go grocery shopping.

### .json file
I've created a file that consists of the fridge's current contents.
Please download the .json file attached in this zip and place it in the same directory as the current .ipynb. 

```python
# some global variables/dialogue
opened = "You have opened the refrigerator.\nThese are your current details.\n"
closed = "\nYou have closed the refrigerator."
options_long = "\n\tInput 'Fridge contents', 'Freezer contents', or 'All' to view groceries.\n\tTo request specific items, input 'Retrieve item 1, item 2, ...'\n\tTo see which items to restock on, input 'Restock'.\n\tTo close the fridge, input 'Close' or 'No'.\n\n"
options_short = "\n\tFridge contents, Freezer contents, or All.\n\tRetrieve item 1, item 2, ...\n\tRestock.\n\tClose the fridge.\n\n"
search = "What would you like to see? Don't worry about capitalization!" + options_long + "\n"
additional = "\nWould you like to look at anything else?" + options_short + "\n"
low = "\nThese items are running low."
groceries = "\nSelect the items that you'd like to add to your grocery list item 1, item 2, ... Input 'All' to add all items.\n"
added = "\nA grocery list has been made.\nPlease refer to the grocery list JSON file in your directory when grocery shopping!\n"

a = "Fridge contents"
b = "Freezer contents"
restock_list = {}
```

```python
import json

def new_input(x): 
    user = input(additional)
    return user
    
def print_key(x):    #prints only key
    print(x)
    for key, value in refrigerator[x].items():
            print(f"\t{key}")
    print("\n")

def print_all(x):    #prints everything
    print(x + "\n")
    for key, value in refrigerator[x].items():
        print(f"\t{key}: {value['quantity']} {value['unit']}")
    print("\n")

def get_restock(x,y,restock_list):
    #prints out contents that need restocking
    print(low)
    for key, value in refrigerator[x].items():
        if value['quantity'] <= 1:
            print(f"\t{key}: {value['quantity']} {value['unit']}")
            restock_list.update({key:""})
    for key, value in refrigerator[y].items():
        if value['quantity'] <= 1:
            print(f"\t{key}: {value['quantity']} {value['unit']}")
            restock_list.update({key:""})

def get_groceries(x,y,restock_list):
    #asks user if they'd like to add items to their grocery list
    og = input("\nWould you like to add items to your grocery list? Yes or No?\n")
    if og.lower() == "yes":
        opt = input(groceries)

        #storing the selected items
        store = {}
        if opt.lower() == "all":
            store = restock_list
            print(store)
        else: 
            opt = opt.split(', ')
            for i in opt: 
                store.update({i:""})

       #output grocery list as a json file
        with open("grocery_list.json", "w") as f:
            json.dump(store, f)
        print(added)  
    else:
        print("\nYou can always add items to your grocery list later.\n")

def list_item(u):     #lists everything of item(s)
   #for every child in that list
    cap_u = [word.capitalize() for word in u]
    
    #finding the parent
    for child in cap_u:
        if child in refrigerator[a].keys():
            parent = a
            for key, value in refrigerator[parent].items():
                if key == child:
                    print(f"\t{key}: {value['quantity']} {value['unit']}")
        elif child in refrigerator[b].keys():
            parent = b
            for key, value in refrigerator[parent].items():
                if key == child:
                    print(f"\t{key}: {value['quantity']} {value['unit']}")
        else: 
            #item doesn't have a parent = not in the fridge
            print(f"\t{child} is not in the fridge.")

    
#check if .json file exists; if yes, fridge is opened
try: 
    file = open("refrigerator_contents.json", "r")
    print(opened)
    
    #read .json file into a python dictionary
    #json.load() = reads a json file and returns a python object
    refrigerator = json.load(file)
    for item in refrigerator:
        print(item)
    print("\n")
except IOError: 
    print("Error: File not found") 

#allow end-user to issue a query and extract a subset of data
user = input(search)

while refrigerator:
    #input valid regardless of upper/lower case
    if user.lower() == "all":
        print_key(a)
        print_key(b)
    elif user.lower() == "fridge contents":
        print_all(a)
    elif user.lower() == "freezer contents":
        print_all(b)
    elif "retrieve" in user.lower():
        #output that item and its quantity
        #allows 2 items and more!
        u = user.lower().replace("retrieve ", "")
        u = u.split(', ')
        list_item(u)    #checking if item(s) in fridge
    elif user.lower() == "restock":
        get_restock(a,b,restock_list)
        get_groceries(a,b,restock_list)
    elif user.lower() == "no" or "close":
        #close the file/fridge
        file.close()
        print(closed)
        break
    else:
        print("\n" + user + " not found. Please ensure input is valid.\n")

    user = new_input(user)
```
